```python
'''    Task 1: DataFrame Basics
     - Display the first five rows
     - Find the number of rows and columns
     - Display all column names and data types  '''

print("First 3 Rows: ")
display(task_df.head())
print("-------------------------------------------")
print("No. of Rows and Columns: ",task_df.shape)
print("-------------------------------------------")
print("Column names and Data Types: ")
display(task_df.dtypes)
```

```python
''' Task 2: Column and Row Selection
- Select only Name and Marks
- Select the first four rows using iloc
- Display the City of the student at index 3 using loc '''
selected_columns=task_df[['Name','Marks']]
display(selected_columns)
print("-------------------------------------------")
print('First 4 Rows: ')
display(task_df.iloc[0:4,:])
print("-------------------------------------------")
display(task_df.loc[3,['City']])
```

```python
''' Task 3: Filtering
- Display students with marks greater than or equal to `80`
- Display students from Lahore or Karachi
- Display Python students whose marks are greater than 80  '''
high_scorers=task_df[task_df['Marks']>=80]
display(high_scorers)
print("---------------------------------------------------------------------------")
print("---------------------------------------------------------------------------")
specific_students=task_df[(task_df['City']=='Lahore')|(task_df['City']=='Karachi')]
display(specific_students)
print("---------------------------------------------------------------------------")
print("---------------------------------------------------------------------------")
python_students=task_df[(task_df['Course']=='Python') & (task_df['Marks']>80)]
display(python_students)
```

```python
''' Task 4: Add and Update Data
- Add a `Bonus_Marks` column equal to 5% of marks
- Update one student's city
- Rename `Fee_Paid` to `Fee_Status`
- Drop the `Bonus_Marks` column after checking it '''
practice_df=task_df.copy()
practice_df['Bonus_Marks']=practice_df['Marks']*0.05

practice_df.loc[2, "City"] = "Rawalpindi"                  # Important & new thing I've learned.

practice_df = practice_df.rename(
    columns={"Fee_Paid":"Fee_Status"})

practice_df=practice_df.drop(columns=['Bonus_Marks'])

display(practice_df)
```

```python
'''   Task 5: Missing Values
- Count missing values in every column
- Fill missing marks with the average marks
- Confirm that no missing marks remain  '''
print(task_df.isnull().sum())
print("--------------------------------")
task_df['Marks']=task_df['Marks'].fillna(task_df['Marks'].median())
display(task_df)
print("---------------------------------------------------------------------------")
print(task_df.notnull().sum())
```

```python
'''   Task 6: Sorting and Aggregations
- Sort students by marks from highest to lowest
- Calculate total, average, minimum, and maximum marks
- Find the number of non-missing marks   '''
sorted_df = task_df.sort_values(by='Marks',ascending=False)
display(sorted_df)
print("---------------------------------------------------------------------------")
print("---------------------------------------------------------------------------")
print("Total marks: ",task_df['Marks'].sum())
print("Total marks: ",task_df['Marks'].mean())
print("Total marks: ",task_df['Marks'].min())
print("Total marks: ",task_df['Marks'].max())
print("----------------------------------")
print("----------------------------------")
print("Non-Missing values: ",task_df['Marks'].isnull().sum())
```

```python
'''      Task 7: GroupBy
- Calculate average marks by course
- Count students by city
- Find the highest marks in every course  '''
course_avg = task_df.groupby('Course')['Marks'].mean()
display(course_avg)
print("--------------------------------")
print("--------------------------------")
student_count=task_df.groupby('City')['Student_ID'].count()
display(student_count)
print("-------------------------------")
print("-------------------------------")
course_max_marks = task_df.groupby('Course')['Marks'].max()
display(course_max_marks)
```

*Level 4: Mini Projects*
16. *CSV Save*: After filtering students with `Age < 18`, save them to a new CSV `minors.csv`.
17. *Merge*: Create 2 DataFrames
`students` and `marks`. Merge them on `Name`.
18. *Rename*: Rename column `City` to `Location`.
19. *Apply*: Create a column `Age_Group` = "Adult" if Age >= 18 else "Minor".
20. *Describe*: Use `.describe()` to get summary stats for numeric columns.

```python
filtered_df = student_df[student_df["Age"]<18]
filtered_df.to_csv("minors.csv",index=False)

students = pd.DataFrame({
    "Name": ["Ali", "Ahmed", "Sara", "Ayesha", "Usman"],
    "Age": [15, 21, 19, 22, 17],
    "City": ["Karachi", "Lahore", "Islamabad", "Karachi", "Peshawar"]
})
marks = pd.DataFrame({
    "Name": ["Ali", "Ahmed", "Sara", "Ayesha", "Usman"],
    "Marks": [85, 92, 78, 88, 95]
})
merge_df=pd.merge(students,marks, on="Name")
merge_df=merge_df.rename(columns={"City":"Location"})
merge_df["Age_Group"]=merge_df["Age"].apply(lambda x:"Adult" if x>= 18 else "Minor")
display(merge_df)
display(merge_df.describe(include="all").round(1))
```
